```python
#Libraries
import nltk
from nltk.corpus import reuters
from nltk import trigrams
from collections import  defaultdict

```

```python
#Download corpus
nltk.download("reuters")
nltk.download("punkt_tab")
```

```python
#Placeholder for model
model = defaultdict(lambda: defaultdict(lambda: 0))  #return 0 for any missing key, store frequencies
```

```python
#Count frequency of 3rd word after the first 2 words in the model
for sentence in reuters.sents():
    for w1, w2, w3 in trigrams(sentence, pad_right=True, pad_left=True): # add padding , to recognize start and end of sentences
        model[(w1, w2)][w3] += 1
```

```python
#Transform freq to probabilities
for w1_w2 in model:
  total_count = float(sum(model[w1_w2].values()))
  for w3 in model[w1_w2]:
    model[w1_w2][w3] /= total_count   # model[w1_w2][w3] = model[w1_w2][w3]/ total_count
```
